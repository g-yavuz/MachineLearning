```python
X_train['annual_milage'] = X_train['milage'] / X_train['car_age'].replace(0, 1)
X_test['annual_milage'] = X_test['milage'] / X_test['car_age'].replace(0, 1)
```

```python
top_ext_colors = X_train["ext_col"].value_counts().head(5)
top_int_colors = X_train["int_col"].value_counts().head(5)
X_train['ext_col_grouped'] = X_train['ext_col'].apply(lambda x: x if x in top_ext_colors else 'Other')
X_train['int_col_grouped'] = X_train['int_col'].apply(lambda x: x if x in top_int_colors else 'Other')
X_test['ext_col_grouped'] = X_test['ext_col'].apply(lambda x: x if x in top_ext_colors else 'Other')
X_test['int_col_grouped'] = X_test['int_col'].apply(lambda x: x if x in top_int_colors else 'Other')
```

```python
X_train = X_train.drop(columns=["ext_col","int_col","engine"],errors="ignore")
X_test = X_test.drop(columns=["ext_col","int_col","engine"],errors="ignore")
```

```python
X_train = pd.get_dummies(X_train,columns=["ext_col_grouped","int_col_grouped"],drop_first=True,dtype=int)
X_test = pd.get_dummies(X_test,columns=["ext_col_grouped","int_col_grouped"],drop_first=True,dtype=int)
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
linear = LinearRegression()
```

```python
linear.fit(X_train_scaled,y_train)
```

```python
y_pred = linear.predict(X_test_scaled)
```

```python
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
score = r2_score(y_test,y_pred)
print("Mean Absolute Error: ", mae)
print("Mean Squared Error: ", mse)
print("R2 Score: ", score)
plt.figure(figsize=(5,4))
plt.scatter(y_test,y_pred)
plt.show()
```
